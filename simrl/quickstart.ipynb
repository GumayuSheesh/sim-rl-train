## Look at the simulator (headless render)

````python
import gymnasium as gym, matplotlib.pyplot as plt
env = gym.make('Ant-v5', render_mode='rgb_array')
env.reset(seed=0)
for _ in range(50):
    env.step(env.action_space.sample())
plt.imshow(env.render()); plt.axis('off'); env.close()
````
